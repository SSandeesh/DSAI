**Object Oriented Programming**
- `Class`: A blueprint or template for creating objects. Defines attributes and behaviors.
- `Object`: An instance of a class. Each object has state (attributes) and behavior (methods).
- `Encapsulation`: Bundling data (attributes) and methods that operate on data within one unit (class), hiding internal state from outside.
- `Inheritance`: Mechanism where a new class derives properties and behaviors from an existing class.
- `Polymorphism`: Ability of different classes to be treated as instances of the same class through inheritance. It allows methods to do different things based on the object calling them.
- `Abstraction`: Hiding complex implementation details and showing only the necessary features.

#### Constructor in Python (`__init__`)
- In user defined method user need to call the function to get output
- (`__init__`) is called **Constructor**

```python
class laptop():
    def __init__(self): # Special method
        print("i5","16GB")
lap1=laptop()
```

```python
class laptop():
    def __init__(self): 
        print("i5","16GB")
lap1=laptop()
lap2=laptop()
```

```python
class laptop():
    def __init__(self): 
        print("i5","16GB")
lap1=laptop()
print()
lap3=laptop()
print()
lap3.__init__()
```

```python
class laptop():
    def __init__(self): 
        print("i5","16GB")
    def config(self):
        print("i7","32GB")
lap1=laptop()
lap1.config()
```

```python
class laptop():
    def __init__(self): 
        print("in it")
    def config(self):
        print("i7","32GB")
lap1=laptop()
lap2=laptop()

lap1.config()
lap2.config()
```

```python
class laptop():
    def __init__(self,cpu): 
        print("in it")
    def config(self):
        print("i7","32GB")
lap1=laptop("i7")
lap2=laptop("i5")

lap1.config()
lap2.config()
```

```python
class laptop():
    def __init__(self,cpu,ram): 
        self.cpu=cpu
        self.ram=ram
        
    def config(self):
        print("config is ", self.cpu, self.ram)
        
lap1=laptop("i7",16)
lap2=laptop("i5",32)

lap1.config()
lap2.config()
```
