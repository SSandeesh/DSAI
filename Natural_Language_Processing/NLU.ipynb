```python
import nltk
```

```python
AI='''Artificial Intelligence refers to the intelligence of machines. This is in contrast to the natural intelligence of
humans and animals. With Artificial Intelligence, machines perform functions such as learning, planning, reasoning and
problem-solving. Most noteworthy, Artificial Intelligence is the simulation of human intelligence by machines.
It is probably the fastest-growing development in the World of technology and innovation. Furthermore, many experts believe
AI could solve major challenges and crisis situations.'''
```

- **word_tokenize()** splits text into words and punctuation.
- Each word is called a token.

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(AI)
words
```

```python
len(words)
```

- **sent_tokenize()** divide text into sentences.

```python
from nltk import sent_tokenize
sentences= sent_tokenize(AI)
sentences
```

- **blankline_tokenize()** splits text into paragraphs based on blank lines.

```python
from nltk.tokenize import blankline_tokenize
AI_blank = blankline_tokenize(AI)
print("Paragraphs:", AI_blank)
```

```python
len(AI_blank)
```

**Stemming** - `normalize words into its base form`
 - porterstemmer
 - lancasterstemmer
 - snowballstemmer

```python
from nltk.stem import PorterStemmer
pst=PorterStemmer()
```

```python
pst.stem("Affection")
```

```python
pst.stem("Playing")
```

```python
pst.stem("Maximum")
```

```python
words_to_stem=["sing","sang","sung","singing"]
for word in words_to_stem:
    print(word + " : " + pst.stem(word))
```

```python
words_to_stem=["Ring","given","gave","singing","running","ate"]
for word in words_to_stem:
    print(word + " : " + pst.stem(word))
```

```python
from nltk.stem import LancasterStemmer
lst=LancasterStemmer()
```

```python
words_to_stem=["Ring","given","gave","singing","running","ate"]
for word in words_to_stem:
    print(word + " : " + lst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sbs=SnowballStemmer("english")
```

```python
words_to_stem=["Ring","given","gave","singing","running","ate"]
for word in words_to_stem:
    print(word + " : " + sbs.stem(word))
```

**Lemmatization**

```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
word_lem=WordNetLemmatizer()
```

```python
words_to_stem
```
