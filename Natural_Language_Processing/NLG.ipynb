## Natural Language Generation

```python
import nltk
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
data="""Data Science, Machine Learning, AI, Big Data, Python, R, SQL, Statistics, 
        Analytics, Visualization, Pandas, NumPy, Matplotlib, TensorFlow, PyTorch, 
        Hadoop, Spark, AWS, Azure, Google Cloud, NLP, Deep Learning, Automation, 
        Prediction, Modeling, Ethics, Privacy, Bias Detection, Dashboards, Insights."""
```

```python
# WordCloud Usage
wordcloud = WordCloud(
    width=400,
    height=200,
    margin=3,
    background_color="black",
    colormap="Accent",
    mode="RGBA"
).generate(data)
```

```python
#plotting WordCloud

plt.imshow(wordcloud,interpolation="quadric")
plt.axis("off")
plt.margins(x=0,y=0)
```
