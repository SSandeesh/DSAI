## Chunking
- `Chunking` refers to divide text into meaningful segments called chunks
- **Techniques**
    - `NP (Noun Phrase)` Groups together nouns and their modifiers
    - `VP (Verb Phrase)` Groups verbs with auxiliary verbs and modifiers	
    - `PP (Prepositional Phrase)` Groups prepositions with their objects
  

```python
import nltk
from nltk import pos_tag, word_tokenize, RegexpParser
```

```python
text="The quick brown fox jumps over the lazy dog"
```

```python
tokens=word_tokenize(text)
tokens
```

```python
tagged_tokens=pos_tag(tokens)
tagged_tokens
```

```python
chunk_grammer=r"""
NP:{<DT>?<JJ>*<NN>}
VP:{<VB.*><NP|PP>*}
PP:{<IN><NP>}
"""
chunk_parser=RegexpParser(chunk_grammer)
chunked=chunk_parser.parse(tagged_tokens)
print(chunked)
chunked.draw()
```
