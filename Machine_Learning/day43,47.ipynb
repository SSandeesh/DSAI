### Machine Learning 
- `🤖 Machine Learning` is a branch of Artificial Intelligence (AI) that gives computers the ability to learn from data and make decisions or predictions without being explicitly programmed.
- **Types of Machine Learning**
    1. **Supervised learning** : Uses labeled data meaning the data has both input features (X) and known output labels (Y). The algorithm learns a mapping from inputs to outputs and predicts the output for new, unseen inputs.
   
        - `Regression` Used when Dependent Variable (Y) is Continuous and Predicts continuous values 
           - eg: Gold Price, Petrol Price, Stock Price Prediction 
        - `Classification` Used when Dependent Variable (Y) is Binary and predict which class (category) new inputs belong to.
           - eg: Profit/Loss, Spam/Not Spam, Customer churn prediction
        
    2. **Unsupervised learning** : Uses unlabeled data only inputs (X), no target (Y). The algorithm finds hidden patterns, structures, or groupings in the data.
        - `Clustering` Groups similar data points
           - eg: Grouping documents, Customer segmentation
- **Model Building Phases**
  - `Training Phase` - Learn patterns from data
  - `Testing Phase` - Evaluate final model accuracy
  - `Validation Phase` - Tune hyperparameters

### Norminal and Ordinal
**Nominal Data** (Name/Label only) - Categories without any order or ranking.Just labels or names. Cannot be sorted.
- Mathematical operations: Not meaningful (e.g., mean, median).
- `Examples`:
    - Gender: Male, Female, Other
    - Blood type: A, B, AB, O
    - Marital Status: Single, Married, Divorced
    - Country names: India, USA, UK
- **Use Case** : When classification is required without any hierarchy.

**Ordinal Data**(Order matters) - Categories with a logical order or ranking.The intervals between categories are not equal.
- Mathematical operations: Mean is not meaningful, but median and mode are.
- `Examples`:
  - Education level: High School < Bachelor < Master < PhD
  - Satisfaction rating: Very Unsatisfied < Unsatisfied < Neutral < Satisfied < Very Satisfied
  - Clothing size: S < M < L < XL
- **Use Case** : When rankings, preferences, or levels need to be represented.

### Underfit and Overfit in ML

- **Underfitting**, A model is said to underfit the data when it is too simple to capture the underlying structure of the data.
    - High bias and low variance
- **Overfitting**, A model is said to overfit the data when it learns not only the underlying pattern but also noise and random fluctuations in the training data.
    - Low bias and high variance
- What are the techniques to reduce `Overfitting`
	1. PCA ( principal component analysis) 
	2. Cross validation 
	3. Regualrization
         - Lasso
         - Ridge     
	4. Ensamble learning 
	5. Drop out the neurons 
	6. Business understanding 

## Confusion Matrix

- TP: Model predicted "Yes", and it was correct.
- FP: Model predicted "Yes", but it was wrong (a false alarm).
- TN: Model predicted "No", and it was correct.
- FN: Model predicted "No", but it was wrong (missed it).
#### **Accuracy**
$${Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$

---
#### **Precision**
 $${Precision} = \frac{TP}{TP + FP}$$

---
#### **Recall** (Sensitivity / True Positive Rate)
$${Recall} = \frac{TP}{TP + FN}$$

---
#### **F1 Score** 
$${F1 Score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$
