### Statistics
- **Population** is the entire group of individuals or items and **Sample** is Subset of population
- **Sampling Techniques**
  - `Random sampling` : randomly selecting samples
  - `Stratified sampling` : selecting samples based on a specific stategy
  - `Systematic sampling` : selecting samples in ordered way (say every 3rd element)
- **Central Tendancy** used to indicate where does middle/centre of the distribution lies. Has three categories:
  - `Mean`- Average
  - `Median`- Middle value after sorting
  - `Mode` - Most repeated value
- **Varience** tells us how far each number in a dataset is from the mean
  
$$StandardVarience = \frac{\sum (x - \bar{x})^2}{N}$$
$$StandardDeviation = \sqrt{StandardVarience}$$
  
- **Range** is the distance b/w maximum and minimum value in data
  - Range = Max - Min
- **Percentile** is a value below which a given percentage of observations fall in a dataset.
  - Suppose 100 students took an exam.If a student is in the 90th percentile, it means 90% of students scored less than or equal to that student.
- **Quartiles** We have three quartiles (Q1, Q2, Q3)
  - `Q1 (25th percentile)` is the value below which 25% of the data falls ,
  - `Q2 (50th percentile)` is the median, the middle of the data
  -  `Q3 (75th percentile)` is the value below which 75% of the data falls
- **Inter Quartile Range(IQR)**
  - `IQR = Q3 - Q1`
- `For Outliers - Boxplot`
- `For Correlation - Heatmap`
### Probability
- **Probability** is the way of measuring how likely something is to happen
  - $Probability=\frac{Favourable Outcomes}{Total Number of Outcomes}$
- Probability is off three types:
  - `Joint Probability P(A,B)`- prob. of events occuring together at the same time
  - `Marginal Probability P(A)`- prob. of event irrespective of another event
  - `Conditional Probability` - prob. of event occuring given that another event had already occured
     - $P(A/B)=\frac{P(A,B)}{P(B)}$
- **Baye's Theorem** Extention of conditional probability
### Probability Distributions
- Prob. of different possible values of a variable 
- Types:
  - Discrete PD
    - `Bernoullis Distribution`- outcome of a event is binary and Number of trails is 1
    - `Binomial Distribution`- Extension of Bernoullis Dist where  outcome of a event is binary and Number of trails is multiple
    - `Poisson's Distribution`- Estimates the likelihood of a event in a given time duration
  - Continuous PD
    - `Normal Distribution`- when distribution follows a bell curve
