#### Descriptive Statistics Analysis
##### `1. Mean, Median, Mode`

```python
print("Mean of the Data : ",df["Population in thousands (2017)"].mean())
print("Median of the Data : ",df["Population in thousands (2017)"].median())
print("Mode of the Data : \n ",df["Population in thousands (2017)"].mode())
```

##### `2. Varience and Standard Deviation`

```python
print("Varience of the Data : ",df["Population in thousands (2017)"].var())
print("Standard Deviation of the Data : ",df["Population in thousands (2017)"].std())
```

##### `3. Range`

```python
maxi = df["Population in thousands (2017)"].max()
mini = df["Population in thousands (2017)"].min()
print("Maximum : ",maxi)
print("Minimum : ",mini)
rang = maxi-mini
print("Range of dataset : ",rang)
```

##### `6. Correlation`
- Correlation is a statistical measure that shows how two numeric variables change with respect to each other.
     - 📈 `Positive correlation`: both increase together
     - 📉 `Negative correlation`: one increases while the other decreases
     - `0 correlation`: no relationship

```python
df.corr(numeric_only=True)
```

```python
# Visualization of Correlation
plt.figure(figsize=(15,10))
sns.heatmap(df.corr(numeric_only=True),annot=True)
```
