### Inferential statistics
- **Descriptive Vs Inferential stats**
  - `Descriptive Stats` - Involves in summarization and presentation of data in a meaningful way.Primary Goal is to describe the data leveraging measures of central Tendency ,measures os shape etc
  - `Inferential Stats` - Involves in making inferences and draws conclusions about a population based on sample of data(random) .Primary goal is to generalize findings from a sample to a larger population leveraging `Hypothesis Testing`,`Confidence Interval`,`T-test`,`Anova` etc
- `Mean` is sensitive to `outliers`
- **Z-score Standardization** is a technique used to rescale data such that the transformed values have a mean of 0 and a standard deviation of 1
$$Z = \frac{(X - \mu)}{\sigma}$$
- **Standard Error** is a measure of how much your sample mean is expected to vary from the true population mean
$$ SE = \frac{s}{\sqrt{n}}$$
- **Margin of Error** is the maximum amount you expect your estimate (like sample mean) to differ from the true population value at a certain confidence level
   $$MoE = Z \times SE$$
- **Confidence interval** is a range of values, derived from sample data, that is likely to contain the true value of a population parameter (like the mean or proportion), with a specified level of confidence
   $$CI = \bar{x} \pm MoE = (\bar{x} - MoE,\ \bar{x} + MoE)$$

### Hypothesis Testing


**Type I Error**(`False Positive - Reject H₀ when it's true`)
- Rejecting the null hypothesis even though it is actually true.
    - Think:  “You said something changed, but it really didn’t.” 
    - Example:  Doctor says you're sick, but you're actually healthy.
      
**Type II Error** (`False Negative - Fail to reject H₀ when it's false`)
- Failing to reject the null hypothesis even though it is actually false.
    - Think:  “You said nothing changed, but something really did.” 
    - Example:  Doctor says you're healthy, but you're actually sick.
