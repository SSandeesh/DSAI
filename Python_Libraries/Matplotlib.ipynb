# Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Scatter Plot when to see correlation b/w variables
xa=np.random.random(50)*100
ya=np.random.random(50)*100
plt.scatter(xa,ya,c="red")
```

```python
#Scatter Plot
xa=np.random.random(50)*100
ya=np.random.random(50)*100
plt.scatter(xa,ya,c="red",marker="+",s=50)
```

```python
years=[2003 + x for x in range(10)]
weights=[34,24,10,45,67,89,9,1,56,55]
plt.plot(years,weights)  # only plot because plot is default
```

```python
# Line plot to show trends over time
years=[2003 + x for x in range(10)]
weights=[34,24,10,45,67,89,9,1,56,55]
plt.plot(years,weights,"g--",marker="*")
```

```python
# Bar Plot when to compare categories
lang=["C++","Python","C#","Java","Go"]
votes=[12,140,20,80,10]
plt.bar(lang,votes)
```

```python
lang=["C++","Python","C#","Java","Go"]
votes=[32,140,20,80,10]
plt.bar(lang,votes,color="r")
```

```python
lang=["C++","Python","C#","Java","Go"]
votes=[12,140,20,80,10]
plt.bar(lang,votes,align="edge",width=0.5, color="g",edgecolor="r")
```

```python
# Histograms when to understand single variable
ages= np.random.random(20)
plt.hist(ages,bins=12)
```

```python
ages= np.random.random(20)
plt.hist(ages,bins=12,cumulative=True)
```

```python
# Pie charts to show parts of a whole
lang=["C++","Python","C#","Java","Go"]
votes=[32,140,20,80,10]
plt.pie(votes,labels=lang)
```

```python
lang=["C++","Python","C#","Java","Go"]
votes=[32,140,20,80,10]
plt.pie(votes,labels=lang)
exp=[0,0,0,0,0.3]
plt.pie(votes,labels=lang,explode=exp,autopct="%.2f%%")
```

```python
# box plot to display summary statistics
heights=np.random.normal(172, 8, 300)
plt.boxplot(heights)
```

```python
years=[2003 + x for x in range(10)]
weights=[5,24,30,45,67,50,89,95,56,100]
wticks=list(range(5,101,10))
plt.plot(years,weights,marker="*")
plt.title("Weights in Years",fontsize=17)
plt.xlabel("Years")
plt.ylabel("Weights")
plt.yticks(wticks,[f"{x}KG" for x in wticks])
```

```python
stk1=[3,22,3,11,3]
stk2=[9,4,34,2,14]
stk3=[55,24,32,12,1]
plt.plot(stk1,label="Company1")
plt.plot(stk2,label="Company2")
plt.plot(stk3,label="Company3")
plt.title("Graph",color="r",fontsize=17)
plt.legend()
```
