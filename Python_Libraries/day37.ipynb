### Tkinter 
- Tkinter is the standard GUI (Graphical User Interface),It allows you to create desktop applications with windows, buttons, text boxes, menus, and more

```python
import tkinter as tk

# create the main application window
root=tk.Tk()
root.title("Simple tkinter App")
root.geometry("200x100")

# Functions to print "Hello World!" in the console
def say_hello():
    print("Hello World!")
    print("Good BYE")

# Create a button that triggers the say_hello() function
hello_button=tk.Button(root,text="Click Me",command=say_hello)
hello_button.pack(pady=40)

# Start the Tkinter event loop 
root.mainloop()
```
