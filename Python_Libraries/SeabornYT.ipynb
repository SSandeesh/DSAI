```python
plt.figure(figsize=(8,6))
```

```python
# 1.scatterplot()
sns.scatterplot(x="total_bill",y="tip",hue="time",data=tips,size="size")
plt.title("Vis of Total Bill vs Tip",color="r",size=18)
```

```python
# 2.lineplot()
sns.lineplot(data=tips,x="size",y="total_bill",hue="sex")
plt.title("Vis of lineplot Size vs total_bill",color="r")
```

```python
sns.lineplot(data=tips,x="size",y="total_bill",hue="sex",ci=False,marker="o")
plt.title("Vis of lineplot Size vs total_bill",color="r")
```

```python
tips.columns
```

```python
# 3.barplot() # hue=group
sns.barplot(x="day",y="total_bill",data=tips,hue="sex",palette="pastel")
plt.title("Vis of Barplot Day vs total_bill",c="r")
```

```python
# 4.Boxplot()
sns.boxplot(data=tips,x="day",y="tip",hue="sex",palette="bright")
plt.title("Vis of Boxplot Day and Tip",c="r")
```

```python
# 5.Violinplot()
sns.violinplot(data=tips,x="day",y="total_bill",hue="time",split=True,palette="muted")
plt.title("Vis of violinplot Day vs total_bill",c="r")
```

```python
# 6.countplot()
sns.countplot(data=tips,x="day",hue="time",palette="dark")
plt.title("Countplot of Day",c="r")
```

```python
# 7.regplot()
sns.regplot(data=tips,x="total_bill",y="tip",scatter_kws={"s":45},line_kws={"color":"r"})
```

```python
# 8.Histogram
sns.histplot(data=tips,x="total_bill",kde=True,color="r",bins=10)
```
