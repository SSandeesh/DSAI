## Anonymous | Lambda Functions

- goalkicker.com

```python
def Square(a):
    return a*a
print(Square(5))
```

```python
f=lambda a:a*a # here anonymous function is called lambda 
f(5)
```

```python
fun=lambda a,b:a+b
res=fun(3,4)
print(f"Addition is {res}")
```

```python
add=lambda a,b:a+b
sub=lambda a,b:a-b
addres=add(1,4)
subres=sub(4,1)
print(f"Addition is {addres}")
print(f"Subtraction is {subres}")
```

#### Lambda has:
        - filter()
        - map()
        - reduce()

### FILTER()

```python
def evens(n):
    return n%2==0
nums=(1,2,3,4,5,6,7,8,9)
even=tuple(filter(evens,nums))
print(even)
```

```python
def odds(n):
    return n%2!=0
nums=[1,2,3,4,5,6,7,8,9]
odd=list(filter(odds,nums))
print(odd)
```

```python
nums=[1,2,3,4,5,6,7,8,9]
evens=list(filter(lambda a:a%2==0,nums))
odds=list(filter(lambda a:a%2!=0,nums))
print(evens)
print(odds)
```

### MAP()

```python
def doublenum(a):
    return a*2
nums = [3,20,6,18,4,6,12,9]
evens=list(filter(lambda a: a%2==0,nums)) 
double=list(map(doublenum,evens))
print(evens)
print(double)
```

```python
nums = [3,20,6,18,4,6,12,9]
evens=list(filter(lambda a:a%2==0,nums))
double=list(map(lambda a: a*2,evens))
print(evens)
print(double)
```

```python
nums = [1,2,3,4,5,6,7,8,9]
evens = list(filter(lambda a:a%2==0, nums))
double = list(map(lambda n : n*2, evens))
add= list(map(lambda n : n+2, evens))
sub= list(map(lambda n : n-2, evens)) 
print(evens)
print(double)  
print(add)
print(sub)
```

### REDUCE()

```python
from functools import reduce
```

```python
nums = [1, 2, 3, 4, 5]
result = reduce(lambda a, b: a + b, nums)
print(result)
```

```python
def multiply(a, b):
    return a * b
nums = [2, 3, 4]
reduce(multiply, nums) 
```

```python
def add(a,b):
    return a+b
nums=[1,5,3,2,8,6,4]
evens=list(filter(lambda a:a%2==0,nums))
triple=list(map(lambda a:a*3,evens))
total=reduce(add,triple)
print(evens)
print(triple)
print(total)
```
